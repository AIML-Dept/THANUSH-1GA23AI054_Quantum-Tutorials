**1.Easy: Implement X, Y, and Z gates individually and print the resulting statevectors**

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)
qc.x(0)

print("Circuit:")
print(qc)

state = Statevector.from_instruction(qc)

print("\nStatevector after X gate:")
print(state)
```

**2. Medium: Construct the |-> state using H and Z**

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)

qc.h(0)
qc.z(0)

print("Circuit:")
print(qc)

state = Statevector.from_instruction(qc)

print("\n|-> Statevector:")
print(state)

expected = np.array([1/np.sqrt(2), -1/np.sqrt(2)])

print("\nExpected:")
print(expected)

print("\nVerification:")
print(np.allclose(state.data, expected))
```

**3.Hard: Apply five random single-qubit gates**

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
import random

gates = ['x', 'y', 'z', 'h', 's']

qc = QuantumCircuit(1)

chosen = random.choices(gates, k=5)

print("Random Gate Sequence:")
print(chosen)

for gate in chosen:
    if gate == 'x':
        qc.x(0)
    elif gate == 'y':
        qc.y(0)
    elif gate == 'z':
        qc.z(0)
    elif gate == 'h':
        qc.h(0)
    elif gate == 's':
        qc.s(0)

print("\nCircuit:")
print(qc)

state = Statevector.from_instruction(qc)

print("\nFinal Statevector:")
print(state)
```

**4. Real-world: Quantum Coin Flip Game**

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)

# Create equal probability
qc.h(0)

print("Quantum Coin Flip Circuit:")
print(qc)

state = Statevector.from_instruction(qc)

print("\nStatevector:")
print(state)

probabilities = state.probabilities()

print("\nProbability of Heads (|0>):", probabilities[0])
print("Probability of Tails (|1>):", probabilities[1])
```
